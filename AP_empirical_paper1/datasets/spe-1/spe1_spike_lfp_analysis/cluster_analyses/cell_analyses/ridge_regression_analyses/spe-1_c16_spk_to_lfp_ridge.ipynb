## Alternative spectral-feature method: single PSD per pre/post window

The targets above (`Y_raw`/`Y_hpf01`/`Y_hpf005`) derive `gamma_auc`/`exponent`/`theta_auc`
by **averaging** an already time-resolved sliding-window Specparam fit
(`specparam_by_spike[i]`) within each pre/post/baseline window.

The cells below build an **alternative** version (`Y_psd`, added — not replacing the above):
for each spike, compute **one PSD** over a segment centered on each window and fit
**one static `SpectralModel`** to it directly, rather than averaging a blend of
overlapping sliding-window epochs. Because the analysis windows here (50-100 ms) are
far too short to estimate a usable PSD across 1-90 Hz on their own (minimum spectral
bandwidth ≈ 1/window_length ≈ 20 Hz), the PSD is computed over a wider 250 ms segment
*centered on* each window's midpoint (`psd_seg_len_s=0.25`) — long enough to resolve
theta (4-8 Hz) separately from gamma (30-55 Hz) while still producing one fit per window.

```python
# ── 5-6. Single-PSD matrices + ridge (skips build+run if pickle exists) ───────
from config import LFP_FS

FORCE_RECOMPUTE_PSD = False  # set True to force rerun even if pickle exists

save_path_psd = os.path.join(RIDGE_DIR, f'c{cell_num}_ridge_results_psd.pkl')

if os.path.exists(save_path_psd) and not FORCE_RECOMPUTE_PSD:
    with open(save_path_psd, 'rb') as f:
        results_psd = pickle.load(f)
    target_names_psd = list(results_psd.keys())
    print(f'Loaded from cache: {os.path.basename(save_path_psd)}')
else:
    (_, _, _, _, Y_psd, target_names_psd, target_labels_psd) = build_ridge_matrices_single_psd(
        df_reg, lfp_windows_by_spike, hpf_lfp_01, LFP_FS,
        pre_win=PRE_WIN, post_win=POST_WIN, baseline_win=BASELINE_WIN,
        psd_seg_len_s=0.25,
    )
    results_psd = run_ridge_regression(
        Y_psd, predictor_sets, target_names_psd,
        n_perm=N_PERM, rng_seed=RNG_SEED, alphas=ALPHAS,
        save_path=save_path_psd, force_recompute=True,
    )
    results_psd = apply_fdr(results_psd, target_names_psd, predictor_sets)
    with open(save_path_psd, 'wb') as f:
        pickle.dump(results_psd, f)
    print(f'Saved: {os.path.basename(save_path_psd)}')
```

## 2. Build Feature and Target Matrices

**Predictors (X)** — three competing sets:
- *Waveform only*: 8 spike shape features
- *Log ISI only*: 1 feature
- *Waveform + Log ISI*: 9 features combined

**Targets (Y)** — 25 LFP scalars (5 groups × 5 features):

| Group | Formula | Scientific question |
|-------|---------|-------------------|
| Pre absolute | mean(pre window) | LFP state when cell fires |
| Pre−BL | mean(pre) − mean(baseline) | LFP ramp into spike |
| Post absolute | mean(post window) | LFP state after spike |
| Post−BL | mean(post) − mean(baseline) | Spike-triggered response from baseline |
| Δ post−pre | mean(post) − mean(pre) | Net spike-triggered change |

Windows: Baseline −200→−100 ms  |  Pre −55→−5 ms  |  Post +5→+55 ms
