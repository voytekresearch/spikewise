```python
import pandas as pd
from pathlib import Path
import numpy as np
import ast

def generate_meta_from_csv(csv_path, bin_file_name, bin_file_dir,
                            sampling_rate=30000, n_channels=384, dtype='int16'):
    # Load the metadata CSV and parse the n_samples
    df = pd.read_csv(csv_path)
    npx_shape = ast.literal_eval(df['npx'].iloc[0])  # (384, n_samples)
    n_samples = npx_shape[1]

    # Compute file size in bytes
    bytes_per_sample = np.dtype(dtype).itemsize
    file_size_bytes = n_channels * n_samples * bytes_per_sample

    # Prepare meta content
    base_name = bin_file_name.replace('.bin', '')
    meta_text = f"""fileName={base_name}
imSampRate={sampling_rate}
nSavedChans={n_channels}
snsSaveChanSubset=0:{n_channels - 1}
fileSizeBytes={file_size_bytes}
dataFormat={dtype}
"""

    # Write the .meta file
    meta_path = Path(bin_file_dir) / f'{base_name}.meta'
    with open(meta_path, 'w') as f:
        f.write(meta_text)

    print(f"✅ Wrote .meta file to: {meta_path}")
    return meta_path

```
